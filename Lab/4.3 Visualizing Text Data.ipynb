```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Netflix_Movies_and_TV_Shows.csv')

df.head()
```

```python
df['Genre'].value_counts().head()
```

```python
# Set visual style
sns.set_style("whitegrid")

# Create Bar Chart for Genres
plt.figure(figsize=(10, 6))
genre_counts = df['Genre'].value_counts().head(10) # Top 10 genres
sns.barplot(x=genre_counts.values, y=genre_counts.index, palette='magma')
plt.title('Top 10 Genres by Frequency')
plt.xlabel('Number of Titles')
plt.show()
```

This horizontal bar chart illustrates the ten most frequent genres in the dataset, where the bar length corresponds to the number of titles. The data shows that Horror and Thriller are the most dominant categories, with each exceeding 390 titles. This highlights a significant preference for suspense-driven content in the library compared to other popular genres like Action or Sci-Fi.

```python
# Create Bar Chart for Countries
plt.figure(figsize=(10, 6))
country_counts = df['Country'].value_counts().head(10)
sns.barplot(x=country_counts.values, y=country_counts.index, palette='viridis')
plt.title('Top 10 Production Countries')
plt.xlabel('Number of Titles')
plt.show()
```

This visualization ranks the top ten countries based on content volume, using a color gradient to differentiate the production output. Canada tops the list with over 400 titles, followed by the United Kingdom and South Korea, while the United States surprisingly ranks fifth. This ranking suggests the dataset may be curated to emphasize international markets rather than U.S.-centric programming.

```python
# Create Bar Chart for Ratings
plt.figure(figsize=(12, 6))
rating_counts = df['Rating'].value_counts()
sns.barplot(x=rating_counts.index, y=rating_counts.values, palette='coolwarm')
plt.title('Distribution of Content Ratings')
plt.ylabel('Count')
plt.xlabel('Rating Category')
plt.show()
```

This vertical bar chart displays the frequency of titles across seven content maturity ratings, such as TV-14, R, and PG. The distribution is fairly balanced among the top categories, though TV-14 and R are the most common. This indicates that the platform's library is geared more toward teen and mature audiences, as G-rated content appears least frequently, but overall all rating seem to be fairly close.
